```python
#importar pandas
import pandas as pd
```

```python
#conjunto de datos base para la generacion del bigdata
datos = {
    "usuario": ["ana", "pepe", "many", "rosy", "carol", "adri"],
    "edad": [14,23,18,40,62,30],
    "ciudad": ["Cuitlahuac", "yanga", "atoyac", "paraje", "cuichapa", "Cosolapa"],
    "compras_realizadas": [5,10,1,2,15,4],
    "activo":[True,False,True,True,False, True],
    "fecha_registro": ["2026-01-10","2026-02-10","2026-02-10","2026-02-10","2026-04-10","2026-04-10"]
}
```

```python
#con los datos creamos el dataframe
dfData = pd.DataFrame(datos)
```

```python
#Mostramo 
dfData
```

```python
#cantidad de registro a generar aleatoriamente
cantidad= 1000000
```

```python
#importar la libreria random
import random
```

```python
#crear el dataframe de 100000 de registros
dfMillon = pd.DataFrame({
    "nombre": [random.choice(dfData["usuario"]) for _ in range(cantidad)],
    "edad": [random.randint(18,80) for _ in range(cantidad)],
    "ciudad": [random.choice(dfData["ciudad"]) for _ in range(cantidad)],
    "compras_realizadas": [random.randint(1,10) for _ in range(cantidad)],
    "activo": [random.choice(dfData["activo"]) for _ in range(cantidad)],
    "fecha_registro": [random.choice(dfData["fecha_registro"]) for _ in range(cantidad)],
})
```
